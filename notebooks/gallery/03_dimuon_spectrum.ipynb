# Dimuon spectrum
**Scope:** synthetic teaching fixture. Cutflow and histogram are generated by HEPLab's public `run` API.

```python
from heplab import labs, run
events = labs.practice_events(4000, seed=2026)
config = {"selection":{"muons":{"min_pt":20,"max_rel_iso":0.15},"pairs":{"require_opposite_charge":True}},"binning":{"dimuon_mass":{"range":[40,140],"bins":100}}}
result = run(events, config=config)
print(result["cutflow"])
print("selected histogram entries:", result["histogram"]["entries"])
```

```python
assert result["histogram"]["entries"] > 0
assert result["provenance"]["event_count"] == 4000
peak_index = max(range(len(result["histogram"]["counts"])), key=result["histogram"]["counts"].__getitem__)
edges = result["histogram"]["edges"]
peak = 0.5 * (edges[peak_index] + edges[peak_index + 1])
assert 86 <= peak <= 96
print(f"Tallest 40--140 GeV bin centre: {peak:.3f} GeV")
```

# First event inspection
**Scope:** fixed synthetic Z→μμ practice fixture. This is an educational sample, not CMS data or detector calibration.

```python
from heplab import labs
from heplab.physics import invariant_mass
events = labs.practice_events(20, seed=2026)
event = next(e for e in events if len(e["Muon"]) >= 2)
print(event["EventInfo"])
print("muons:", event["Muon"][:2])
```

```python
mu1, mu2 = event["Muon"][:2]
mass = invariant_mass(mu1, mu2)
assert mass > 0
print(f"First two-muon invariant mass: {mass:.3f} GeV/c^2")
```

## Parameter units
- `pt`: GeV/c
- `eta`, `phi`: dimensionless
- `mass`: GeV/c²
- invariant mass: GeV/c²

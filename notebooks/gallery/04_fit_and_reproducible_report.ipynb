# Fit and reproducible report
**Scope:** synthetic Z teaching sample. The Gaussian fit is an educational model, not a CMS measurement.

```python
import tempfile, pathlib
import heplab
events = heplab.labs.practice_events(4000, seed=2026)
config = {"selection":{"muons":{"min_pt":20,"max_rel_iso":0.15},"pairs":{"require_opposite_charge":True}},"binning":{"dimuon_mass":{"range":[40,140],"bins":100}},"candidate_policy":"all_pairs","validation":{"mode":"educational"}}
result = heplab.run(events, config=config)
try:
 fit = heplab.fit_peak(result, window=(70,110))
except ImportError as exc:
 fit = {"success":False,"reason":str(exc)}
print(fit)
```

```python
assert result["provenance"]["event_count"] == 4000
assert result["result_sha256"]
with tempfile.TemporaryDirectory() as folder:
 path = pathlib.Path(folder)/"z-result.json"
 path.write_text(__import__("json").dumps(result), encoding="utf-8")
 loaded = heplab.load(path)
 assert loaded["result_sha256"] == result["result_sha256"]
print("Artifact digest reload passed")
```

## Completion criteria
- The replay uses the public API.
- The in-memory result and reloaded digest agree.
